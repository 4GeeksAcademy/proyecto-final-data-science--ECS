# CalCOFI-Investigaciones Cooperativas de Pesca Oceánica de California
Datos oceanográficos

-Un poco de historia

Este conjunto comenzó en el año 1949 y continúa en el presente recopilando una serie de datos hidrográficos, físicos, químicos y biológicos tanto en lugares delimitados como durante el trayecto. Recolectados a intervalos regulares de tiempo y espacio se convirtieron rápidamente en valiosos para documentar los ciclos climáticos en la Corriente de California y diversas respuestas biológicas a éstos. La investigación de CalCOFI atrajo la atención mundial sobre la respuesta biológica al drástico calentamiento del Pacífico en 1957-58 e introdujo el término "El Niño" en la literatura científica. 



# Dentro

• Con más de 50 000 estaciones de muestreo, se han lanzado al mar botellas a profundidades de hasta 500 m que recolectaron variables como: 

Temperatura, salinidad, oxígeno, fosfato, silicato, nitrato y nitrito, clorofila, transmitancia, PAR, productividad primaria C14, biodiversidad del fitoplancton, biomasa del zooplancton y biodiversidad del zooplancton.

• Incluye información sobre la abundancia de larvas de más de 250 especies de peces y huevos de especies comerciales clave

Tenemos por lo tanto 3 Dataset con información:

- Bottle: el estudio con botella Niskin utiliza este cilindro especial de PVC para capturar muestras de agua a diferentes profundidades sin contaminarlas
- Cast:  en qué lugar ocurrieron estos lanzamientos, hora fecha y otras condiciones climatológicas
- Biologico:  Sobre biodiversidad marina en esos lugares

# Pasando todo a base de datos sqlite


```python
# Pasando todo a base de datos sqlite
# 1ro Bottle

import sqlite3
import pandas as pd

botellas= "../data/raw/bottle.csv"
bottle = "bottle.db"
las_botellas = "bottle"

#  Leer el archivo CSV con Pandas
bottle = pd.read_csv(botellas)

#  Conectar a SQLite (Crea el archivo .db si no existe)
conexion = sqlite3.connect("bottle.db")

# Guardar el DataFrame en la base de datos como una tabla
bottle.to_sql(las_botellas, conexion, if_exists="replace", index=False)


print(f"¡Base de datos '{bottle}' creada con éxito con la tabla '{las_botellas}'!")
```

```python
# Pasando todo a base de datos sqlite
# 2do Cast

import sqlite3
import pandas as pd

muestras = "../data/raw/cast.csv"
cast = "cast.db"
las_muestras = "cast"

#  Leer el archivo CSV con Pandas
cast = pd.read_csv(muestras)

#  Conectar a SQLite (Crea el archivo .db si no existe)
conexion = sqlite3.connect("cast.db")

# Guardar el DataFrame en la base de datos como una tabla
cast.to_sql(las_muestras, conexion, if_exists="replace", index=False)


print(f"¡Base de datos '{cast}' creada con éxito con la tabla '{las_muestras}'!")
```

Seguimos teniendo los archivos originales, hay que sobreesribirlos

```python
import sqlite3
import pandas as pd

# 1. Leer el CSV original (o cargar la tabla desde SQLite)
bottle = pd.read_csv("../data/raw/bottle.csv", low_memory=False)

# 2. Filtrar columnas con MENOS del 80% de nulos
porcentaje_nulos = bottle.isnull().mean()
cols_utiles_bottle = porcentaje_nulos[porcentaje_nulos < 0.80].index
df_bottle_limpio = bottle[cols_utiles_bottle]

# 3. Sobrescribir la base de datos bottle.db
conn = sqlite3.connect('bottle.db')
bottle_limpio.to_sql('bottle', conn, if_exists='replace', index=False)

# 4. Verificar el número final de columnas guardadas
cols_guardadas = pd.read_sql_query("PRAGMA table_info(bottle);", conn)['name'].tolist()
conn.close()

print(f"Columnas originales en bottle: {bottle.shape[1]}")
print(f"Columnas guardadas en 'bottle.db': {len(cols_guardadas)}")
```

Ahora si, qué columnas finales nos quedaron en Bottle y en Cast

```python
import sqlite3
import pandas as pd
import re

# 1. Leer el archivo Markdown y construir el diccionario de descripciones
ruta_markdown = "/workspaces/proyecto-final-data-science--ECS/documentos/diccionario_variables_calcofi.md"

diccionario_md = {}
with open(ruta_markdown, "r", encoding="utf-8") as f:
    for linea in f:
        # Busca patrones tipo: - **NombreVariable**: Descripción  o  | NombreVariable | Descripción |
        match = re.search(r"[*|`]?([A-Za-z0-9_]+)[*|`]?\s*[:|-|\|]\s*(.*)", linea)
        if match:
            var_nombre = match.group(1).strip()
            var_desc = match.group(2).strip().strip("|").strip()
            if var_nombre and var_desc:
                diccionario_md[var_nombre] = var_desc

# 2. Obtener las variables actuales de bottle.db
conn_b = sqlite3.connect("bottle.db")
vars_bottle = pd.read_sql_query("PRAGMA table_info(bottle);", conn_b)['name'].tolist()
conn_b.close()



# --- IMPRIMIR BOTTLE CON DICCIONARIO MD ---
print(f" VARIABLES EN 'bottle.db' ({len(vars_bottle)} en total):")
print("=" * 80)
for i, var in enumerate(vars_bottle, 1):
    desc = diccionario_md.get(var, "Descripción no encontrada en el .md")
    print(f"{i:2d}. {var:<12} -> {desc}")

print("\n" + "=" * 80 + "\n")

```

```python
# 3. Obtener las variables actuales de cast.db
conn_c = sqlite3.connect("cast.db")
vars_cast = pd.read_sql_query("PRAGMA table_info(cast);", conn_c)['name'].tolist()
conn_c.close()

# --- IMPRIMIR CAST CON DICCIONARIO MD ---
print(f" VARIABLES EN 'cast.db' ({len(vars_cast)} en total):")
print("=" * 80)
for i, var in enumerate(vars_cast, 1):
    desc = diccionario_md.get(var, "Descripción no encontrada en el .md")
    print(f"{i:2d}. {var:<12} -> {desc}")
```
